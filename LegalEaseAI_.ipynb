```python
# ================================
# LegalEase - ONE CELL COLAB APP
# ================================

!pip -q install flask

import threading
import time
import socket
from flask import Flask, request, render_template_string
from google.colab import output

app = Flask(__name__)

HTML = """
<!DOCTYPE html>
<html>
<head>
<meta name="viewport" content="width=device-width,initial-scale=1">
<title>LegalEase</title>
<style>
body {
    font-family: Arial, sans-serif;
    background: #f4f6f8;
    margin: 0;
    padding: 20px;
    color: #17202a;
}
.container {
    max-width: 900px;
    margin: auto;
}
.card {
    background: white;
    padding: 25px;
    border-radius: 16px;
    box-shadow: 0 4px 20px rgba(0,0,0,.08);
    margin-bottom: 20px;
}
h1 { margin-top: 0; }
label {
    display: block;
    font-weight: bold;
    margin-top: 15px;
}
input, select, textarea {
    width: 100%;
    box-sizing: border-box;
    padding: 12px;
    margin-top: 6px;
    border: 1px solid #ccd3da;
    border-radius: 8px;
    font-size: 16px;
}
textarea { min-height: 130px; }
button {
    margin-top: 20px;
    padding: 13px 20px;
    border: 0;
    border-radius: 8px;
    background: #17202a;
    color: white;
    font-size: 16px;
    cursor: pointer;
}
.document {
    white-space: pre-wrap;
    line-height: 1.7;
    font-family: Georgia, serif;
    border: 1px solid #ddd;
    padding: 25px;
    background: white;
    margin-top: 20px;
}
.back {
    display: inline-block;
    margin-top: 20px;
}
.small { color: #68737d; }

@media print {
    body { background: white; }
    .no-print { display: none; }
    .card { box-shadow: none; }
}
</style>
</head>

<body>
<div class="container">

<div class="card">

<h1>⚖  LegalEase</h1>
<p class="small">
AI-powered legal document drafting platform
</p>

<form method="POST">

<label>Document Type</label>
<select name="document_type">
<option value="employment">Employment Contract</option>
<option value="nda">Non-Disclosure Agreement</option>
<option value="lease">Residential Lease Agreement</option>
</select>

<label>First Party</label>
<input name="party1"
placeholder="Company / Employer / Landlord / Disclosing Party"
required>

<label>Second Party</label>
<input name="party2"
placeholder="Employee / Tenant / Receiving Party"
required>

<label>Effective Date</label>
<input type="date" name="date" required>

<label>Additional Terms</label>
<textarea name="terms"
placeholder="Salary, job role, property address, confidentiality scope, notice period, special clauses, etc."></textarea>

<button type="submit">
Generate Legal Document
</button>

</form>

</div>
</div>
</body>
</html>
"""

RESULT_HTML = """
<!DOCTYPE html>
<html>
<head>
<meta name="viewport" content="width=device-width,initial-scale=1">
<title>LegalEase Document</title>

<style>
body {
    font-family: Arial, sans-serif;
    background: #f4f6f8;
    margin: 0;
    padding: 20px;
}
.container {
    max-width: 900px;
    margin: auto;
}
.card {
    background: white;
    padding: 25px;
    border-radius: 16px;
    box-shadow: 0 4px 20px rgba(0,0,0,.08);
}
.document {
    white-space: pre-wrap;
    font-family: Georgia, serif;
    line-height: 1.7;
    padding: 30px;
    border: 1px solid #ddd;
    margin-top: 20px;
    background: white;
}
button {
    padding: 12px 18px;
    border: 0;
    border-radius: 8px;
    background: #17202a;
    color: white;
    font-size: 16px;
}
a {
    display: inline-block;
    margin-top: 20px;
}
@media print {
    body { background: white; }
    .no-print { display: none; }
    .card { box-shadow: none; }
}
</style>
</head>

<body>

<div class="container">
<div class="card">

<div class="no-print">
<h1>⚖  LegalEase</h1>
<button onclick="window.print()">
Print / Save as PDF
</button>
</div>

<div class="document">{{ document }}</div>

<div class="no-print">
<a href="/">← Create another document</a>
</div>

</div>
</div>

</body>
</html>
"""


def generate_document(doc_type, party1, party2, date, terms):

    terms = terms or "No additional terms were specified."

    if doc_type == "employment":

        return f"""
EMPLOYMENT CONTRACT

Effective Date: {date}

EMPLOYER:
{party1}

EMPLOYEE:
{party2}


1. POSITION AND RESPONSIBILITIES

The Employee agrees to perform the duties and responsibilities assigned by the Employer and to perform such duties professionally and in accordance with applicable policies.


2. COMPENSATION

The Employee's compensation and payment arrangements shall be determined and agreed upon by the parties.


3. CONFIDENTIALITY

The Employee shall maintain the confidentiality of non-public business, technical, financial, customer, and operational information obtained through employment.


4. TERM AND TERMINATION

The employment relationship shall continue according to the agreed terms and applicable law.


5. INTELLECTUAL PROPERTY

Work created within the scope of employment shall be handled according to the parties' agreement and applicable law.


6. ADDITIONAL TERMS

{terms}


SIGNATURES


Employer: ______________________________

Name: {party1}

Date: __________________


Employee: ______________________________

Name: {party2}

Date: __________________
"""

    elif doc_type == "nda":

        return f"""
NON-DISCLOSURE AGREEMENT

Effective Date: {date}

DISCLOSING PARTY:
{party1}

RECEIVING PARTY:
{party2}


1. PURPOSE

The parties may exchange confidential information for a legitimate business or professional purpose.


2. CONFIDENTIAL INFORMATION

Confidential Information may include non-public business, financial, technical, commercial, customer, product, or operational information.


3. CONFIDENTIALITY OBLIGATIONS

The Receiving Party agrees to use Confidential Information only for the agreed purpose and to take reasonable measures to prevent unauthorized disclosure.


4. EXCLUSIONS

Information that is publicly available, independently developed, lawfully received from another source, or required to be disclosed by law may be excluded where applicable.


5. TERM

Confidentiality obligations shall continue for the period agreed by the parties and permitted by applicable law.


6. ADDITIONAL TERMS

{terms}


SIGNATURES


Disclosing Party: ______________________________

Name: {party1}

Date: __________________


Receiving Party: ______________________________

Name: {party2}

Date: __________________
"""

    else:

        return f"""
RESIDENTIAL LEASE AGREEMENT

Effective Date: {date}

LANDLORD:
{party1}

TENANT:
{party2}


1. PROPERTY

The Landlord agrees to lease the residential property described by the parties under this Agreement.


2. RENT

The Tenant agrees to pay rent and other agreed charges according to the payment schedule established by the parties.


3. LEASE TERM

The lease shall begin on the effective date and continue for the period agreed by the parties.


4. USE OF PROPERTY

The property shall be used for lawful residential purposes unless otherwise agreed.


5. MAINTENANCE

The parties shall be responsible for maintenance, repairs, and utilities according to their agreement and applicable law.


6. ADDITIONAL TERMS

{terms}


SIGNATURES


Landlord: ______________________________

Name: {party1}

Date: __________________


Tenant: ______________________________

Name: {party2}

Date: __________________
"""


@app.route("/", methods=["GET"])
def home():
    return render_template_string(HTML)


@app.route("/", methods=["POST"])
def generate():

    doc_type = request.form.get("document_type")
    party1 = request.form.get("party1")
    party2 = request.form.get("party2")
    date = request.form.get("date")
    terms = request.form.get("terms")

    document = generate_document(
        doc_type,
        party1,
        party2,
        date,
        terms
    )

    return render_template_string(
        RESULT_HTML,
        document=document
    )


# Dynamically find a free port to avoid "Address already in use" errors
def find_free_port():
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    s.bind(('localhost', 0))
    port = s.getsockname()[1]
    s.close()
    return port

port = find_free_port()

def run_app():
    app.run(
        host="0.0.0.0",
        port=port,
        debug=False,
        use_reloader=False
    )


# Start Flask
thread = threading.Thread(target=run_app)
thread.daemon = True
thread.start()

time.sleep(2)

print("\n" + "=" * 60)
print("   LEGALEASE IS LIVE")
print("=" * 60)
print(f"\n  APP IS RUNNING ON PORT {port}. EMBEDDING COMPATIBLE VIEW BELOW:")
output.serve_kernel_port_as_iframe(port, height=800)
print("=" * 60)
```
